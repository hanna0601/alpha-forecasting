```python
import pandas as pd
import glob

# Folder where your core parquet files are stored
DATA_DIR = "../data/market/"

# Find all yearly price files
files = sorted(glob.glob(f"{DATA_DIR}/prices_*.parquet"))

print("Found files:", files)

dfs = []

for f in files:
    print(f"Reading {f} ...")
    df = pd.read_parquet(f)
    dfs.append(df)

# Combine all years
full_df = pd.concat(dfs, ignore_index=True)

# Optional: sort by Date, Ticker
full_df = full_df.sort_values(["Date", "Ticker"]).reset_index(drop=True)

# Save as CSV
output_csv = f"{DATA_DIR}/all_prices_2014_2023.csv"
full_df.to_csv(output_csv, index=False)

print(f"\nSaved combined CSV → {output_csv}")
print(f"Total rows: {len(full_df):,}")
print(f"Tickers: {full_df['Ticker'].nunique()}")
```
