```python
import yfinance as yf
import pandas as pd
import numpy as np
from tqdm import tqdm
```

```python
sp500 = pd.read_csv("../data/sp500_companies.csv")
```

```python
tickers = sorted(sp500['ticker'].unique())
```

```python
def download_prices(tickers, start="2014-01-01", end="2024-01-01"):
    data = yf.download(
        tickers,
        start=start,
        end=end,
        auto_adjust=True,      # adjusts for splits & dividends
        group_by='ticker',
        threads=True
    )
    return data
```

```python
prices = download_prices(
    tickers,
    start="2014-01-01",
    end="2024-01-01"
)
```
